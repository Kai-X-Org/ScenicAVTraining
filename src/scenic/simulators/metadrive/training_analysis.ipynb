```python
import numpy as np
import pickle
import matplotlib.pyplot as plt
```

```python
# training baseline, model is model/ppo_random_train.pth
train_log = np.genfromtxt("train_logs/uniform_training_start_point.txt", dtype="str")
timestep, epi_avg_reward = train_log[:, 2].astype(np.float32), train_log[:, 11].astype(np.float32)
plt.figure()
plt.xlabel("Timesteps")
plt.ylabel("Avg Episode Reward (Last 100 episodes)")
plt.plot(timestep, epi_avg_reward)
plt.show()
```

```python
# uniform_deviation_baseline: continued uniform random training of the start point
# training baseline, model is model/ppo_random_train.pth
train_log = np.genfromtxt("train_logs/uniform_deviation_baseline_05190233.txt", dtype="str")
timestep, epi_avg_reward = train_log[:, 3].astype(np.float32), train_log[:, 12].astype(np.float32)
plt.figure()
plt.xlabel("Timesteps")
plt.ylabel("Avg Episode Reward (Last 100 episodes)")
plt.plot(timestep, epi_avg_reward)
plt.show()
```

```python
# Eval results: uniform continued trainnig:
filename = 'eval_results/uniform_deviation_baseline_05_19_02_33.pkl'

with open(filename, 'rb') as file:
    loaded_dict = pickle.load(file)
    
episode_reward = loaded_dict["episode_rewards"]
mean_reward = np.mean(episode_reward)
stddev_reward = np.std(episode_reward)

print(f"Reward Mean: {mean_reward}")
print(f"Reward StdDev: {stddev_reward}")
```
